```python
# Write Monk scale from https://skintone.google/get-started website into dictionary
monk_scale = {
    1: '#f6ede4',
    2: '#f3e7db',
    3: '#f7ead0',
    4: '#eadaba',
    5: '#d7bd96',
    6: '#a07e56',
    7: '#825c43',
    8: '#604134',
    9: '#3a312a',
    10: '#292420'
}
```

```python
# dictionary → DataFrame with two named columns
monk_df = pd.DataFrame(list(monk_scale.items()), columns=['monk_tone', 'hex'])

# save; index=False stops pandas adding an extra 0–9 numbering column
monk_df.to_csv('../data/raw/monk_scale.csv', index=False)
```
