## Testing merge options with different keys

Matching strategy: step-by-step (waterfall) matching  
- Merge 1 (brand + product name) matches what it can; Merge 2 (product ID) only tries what's left (no_price); Merge 3 (fuzzy matching) only tries what's still left (still_no_price)  
- After each merge, the matched shop products are removed from the pools (ulta_pool, sephora_pool), so no shop product is used twice  
- Each product gets exactly one match, and the final count is easy: Merge 1 + Merge 2 + Merge 3 = products with a price  
- Key order: best practice is the most reliable key first (usually an exact ID). I started with brand + product name as the most intuitive key and to practise normalizing; the ID key's strength only showed up later. Kept the order, since it mostly changes which key gets the credit, not how many products match. Possible improvement: ID key first  

Python decides, dbt applies:  
- The merges here build a mapping table: one row per product with my product's url, the matched shop product, the source (Ulta or Sephora) and the merge used  
- The mapping table goes into the warehouse like a seed; in dbt, the join is then one simple join: shades → mapping → prices

### Merge 3 - fuzzy matching

to see if I can improve the match rate I want to use fuzzy matching, as sometimes the names of a product are not 100% the

```python
from rapidfuzz import fuzz
```

```python
# compare two texts with fuzz: 100 = identical, 0 = nothing in common
# fuzz.ratio compares the texts character by character, so word order matters.
fuzz.ratio('vanish seamless finish foundation stick', 'vanish seamless finish foundation')
```

```python
# fuzz.token_sort_ratio sorts the words alphabetically first, so “foundation stick vanish” and “vanish foundation stick” score 100. 
# That’s handy when brands shuffle words around.
fuzz.token_sort_ratio('foundation stick vanish', 'vanish foundation stick')
```

```python
from rapidfuzz import process

# an empty list (container): it will collect one result (one small dictionary) per unmatched product, becoming a list of dictionaries
results = []

# go through the unmatched products one at a time
# iterrows() gives two things per row: the index (not needed, so "_") and the full row itself
for _, row in still_no_price.iterrows():

    # the candidates for THIS product: only Ulta pool products of the same brand
    candidates = ulta_pool_2[ulta_pool_2['brand_norm'] == row['brand_norm']]

    # if Ulta has no product of this brand at all, there's nothing to compare
    if candidates.empty:
        # record the product anyway, with empty (None) matches, scores and link
        results.append({'url': row['url'], 'product_norm': row['product_norm'],
                        'match_ratio': None, 'score_ratio': None,
                        'match_token_sort': None, 'score_token_sort': None,
                        'link_token_sort': None})                                   # NEW
        # skip the rest of the loop for this product, and go to the next one
        continue

    # find the single most similar candidate name, with each scorer
    # extractOne returns three things: (matched name, score, position)
    best_ratio = process.extractOne(row['product_norm'], candidates['product_norm'], scorer=fuzz.ratio)
    best_sort = process.extractOne(row['product_norm'], candidates['product_norm'], scorer=fuzz.token_sort_ratio)

    # add this product's result to the list
    # [0] = the matched name, [1] = the score
    # [2] = the position: used to fetch the Ulta link of the token_sort match, needed later for its price
    results.append({'url': row['url'], 'product_norm': row['product_norm'],
                    'match_ratio': best_ratio[0], 'score_ratio': best_ratio[1],
                    'match_token_sort': best_sort[0], 'score_token_sort': best_sort[1],
                    'link_token_sort': candidates.loc[best_sort[2], 'product_link']})  # NEW

# after the loop: turn the list of results into a table
fuzzy_ulta = pd.DataFrame(results)
```

it seems that I can improve the merge score a little by manually checking the result of the fuzzy match

```python
# review_ulta: a NEW DataFrame for the manual review
# it puts my product and its fuzzy candidate side by side, so I can decide if they're the same
# built from fuzzy_ulta (one row per product in still_no_price, 167 rows) with two merges:
review_ulta = (fuzzy_ulta
    # merge 1: add MY product's readable columns (original brand and product name, retailer)
    # matched on url, my product's name tag
    .merge(still_no_price[['url', 'brand', 'product', 'retailer']], on='url', how='left')
    # merge 2: add the ULTA candidate's details (real name, brand, price, description)
    # matched on link_token_sort (the candidate's link from the loop) = Ulta's product_link
    # suffixes: my columns keep their names, Ulta's get '_ulta' (brand -> brand_ulta)
    .merge(ulta_pool_2[['product_link', 'brand', 'product_name', 'price', 'description']],
           left_on='link_token_sort', right_on='product_link', how='left',
           suffixes=('', '_ulta')))

# check: still 167 rows (one per product), nothing doubled
review_ulta.shape
```

```python
# empty column for my decision (yes / no) during the manual review
review_ulta['accepted'] = ''

# sorted by score, so the most likely matches come first
review_ulta.sort_values('score_token_sort', ascending=False).to_csv('../data/manual/review_ulta.csv', index=False)
```

```python
# fuzzy sephora
# an empty list (container): it will collect one result (one small dictionary) per unmatched product, becoming a list of dictionaries
results = []

# go through the unmatched products one at a time
# iterrows() gives two things per row: the index (not needed, so "_") and the full row itself
for _, row in still_no_price_2.iterrows():

    # the candidates for THIS product: only Sephora pool products of the same brand
    candidates = sephora_pool_2[sephora_pool_2['brand_norm'] == row['brand_norm']]

    # if Sephora has no product of this brand at all, there's nothing to compare
    if candidates.empty:
        # record the product anyway, with empty (None) matches, scores and id
        results.append({'url': row['url'], 'product_norm': row['product_norm'],
                        'match_ratio': None, 'score_ratio': None,
                        'match_token_sort': None, 'score_token_sort': None,
                        'id_token_sort': None})
        # skip the rest of the loop for this product, and go to the next one
        continue

    # find the single most similar candidate name, with each scorer
    # extractOne returns three things: (matched name, score, position)
    best_ratio = process.extractOne(row['product_norm'], candidates['product_norm'], scorer=fuzz.ratio)
    best_sort = process.extractOne(row['product_norm'], candidates['product_norm'], scorer=fuzz.token_sort_ratio)

    # add this product's result to the list
    # [0] = the matched name, [1] = the score
    # [2] = the position: used to fetch the Sephora product_id of the token_sort match, needed later for its price
    results.append({'url': row['url'], 'product_norm': row['product_norm'],
                    'match_ratio': best_ratio[0], 'score_ratio': best_ratio[1],
                    'match_token_sort': best_sort[0], 'score_token_sort': best_sort[1],
                    'id_token_sort': candidates.loc[best_sort[2], 'product_id']})

# after the loop: turn the list of results into a table
fuzzy_sephora = pd.DataFrame(results)
```

## Building the mapping table

Purpose: a list of matching decisions, one row per matched product: which shades product belongs to which shop product. dbt later uses it to join the prices.  

Steps:  
- take the matches from all three merges (Merge 1: name, Merge 2: ID, Merge 3: accepted fuzzy matches)  
- give each part the same five columns  
- stack them into one table with pd.concat  

| Column | What it is | Where it comes from |
| --- | --- | --- |
| url | the shades product's name tag: its link, unique per product | the shades side of each merge result (from shades_products) |
| shop_key | the shop product's name tag: product_link (Ulta) or product_id (Sephora) | the shop side of each merge result, or link_token_sort / id_token_sort from the review files |
| source | which shop: ulta or sephora | added with .assign() |
| method | which merge found the match: name, id or fuzzy | added with .assign() |
| successor | flag for renamed or reformulated successor products | the review files (fuzzy matches only; empty for Merges 1 and 2) |

In dbt: shades (url) → mapping (shop_key, source) → ulta or sephora table (shop_key) → price

```python
# one small table per merge and source, all with the same columns
parts = [
    # Merge 1: brand + product name
    merge_ulta_brand_product.loc[merge_ulta_brand_product['_merge'] == 'both', ['url', 'product_link']]
        .rename(columns={'product_link': 'shop_key'}).assign(source='ulta', method='name'),
    merge_sephora_brand_product.loc[merge_sephora_brand_product['_merge'] == 'both', ['url', 'product_id']]
        .rename(columns={'product_id': 'shop_key'}).assign(source='sephora', method='name'),

    # Merge 2: product ID
    merge_ulta_product_id.loc[merge_ulta_product_id['_merge'] == 'both', ['url', 'product_link']]
        .rename(columns={'product_link': 'shop_key'}).assign(source='ulta', method='id'),
    merge_sephora_product_id.loc[merge_sephora_product_id['_merge'] == 'both', ['url', 'product_id']]
        .rename(columns={'product_id': 'shop_key'}).assign(source='sephora', method='id'),

    # Merge 3: fuzzy matching (accepted rows of the reviews), with the successor flag
    ulta_accepted[['url', 'link_token_sort', 'successor']]
        .rename(columns={'link_token_sort': 'shop_key'}).assign(source='ulta', method='fuzzy'),
    sephora_accepted[['url', 'id_token_sort', 'successor']]
        .rename(columns={'id_token_sort': 'shop_key'}).assign(source='sephora', method='fuzzy'),
]

# stack them into one table
mapping = pd.concat(parts, ignore_index=True)

mapping.shape
```
